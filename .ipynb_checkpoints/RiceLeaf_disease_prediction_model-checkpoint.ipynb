# Importing Libraries

```python
# Import necessary libraries
import os
import cv2
import numpy as np
import pandas as pd
import numpy as np
import zipfile
import urllib.request
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, GlobalAveragePooling2D
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.applications import MobileNetV2
from sklearn.metrics import accuracy_score, precision_recall_fscore_support,classification_report,confusion_matrix
from tensorflow.keras.callbacks import EarlyStopping,ModelCheckpoint
```

# Datasset

```python
dataset_url = "https://d3ilbtxij3aepc.cloudfront.net/projects/CDS-Capstone-Projects/PRCP-1001-RiceLeaf.zip"
zip_path = "PRCP-1001-RiceLeaf.zip"

# Downloading and extracting dataset 
if not os.path.exists(zip_path):
    print("Downloading dataset...")
    urllib.request.urlretrieve(dataset_url, zip_path)
    print("Download complete!")


if not os.path.exists('./Data'):
    print("Extracting dataset...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall("./")
    print("Extraction complete!")

DATASET_PATH = './Data'
CLASSES = ['Bacterial leaf blight', 'Brown spot', 'Leaf smut']

# Find and extract all inner zip files inside the Data folder
for item in os.listdir(DATASET_PATH):
    if item.endswith('.zip'):
        zip_file_path = os.path.join(DATASET_PATH, item)
        print(f"Extracting: {item}...")
        with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
            zip_ref.extractall(DATASET_PATH)

print("\nExtraction complete! Here are the folders now available inside Data:")
print(os.listdir(DATASET_PATH))
```

# Exploratory Data Analysis (EDA) 

```python
# 1. Check Class Distribution
class_counts = {cls: len(os.listdir(os.path.join(DATASET_PATH, cls))) for cls in CLASSES}

print("--- Class Distribution ---")
for cls, count in class_counts.items():
    print(f"{cls}: {count} images")

# Plot distribution
plt.figure(figsize=(7, 4))
sns.barplot(
    x=list(class_counts.keys()), 
    y=list(class_counts.values()), 
    hue=list(class_counts.keys()), 
    palette='viridis', 
    legend=False
)
plt.title('Distribution of Rice Leaf Disease Classes')
plt.ylabel('Number of Images')
plt.show()

```

```python
# 2. Visualize Sample Images from Each Class
plt.figure(figsize=(12, 4))
for i, cls in enumerate(CLASSES):
    cls_path = os.path.join(DATASET_PATH, cls)
    img_name = os.listdir(cls_path)[0]
    img = cv2.imread(os.path.join(cls_path, img_name))
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    
    plt.subplot(1, 3, i + 1)
    plt.imshow(img)
    plt.title(f"{cls}\nShape: {img.shape}")
    plt.axis('off')
plt.tight_layout()
plt.show()
```

# Data Augmentation & Preprocessing Pipeline

```python
IMAGE_SIZE = (224, 224)
BATCH_SIZE = 16

# Configure Data Augmentation to prevent overfitting on the 120-image dataset
train_datagen = ImageDataGenerator(
    rescale=1.0 / 255.0,
    rotation_range=35,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest',
    validation_split=0.2  # 80% Training (96 images), 20% Validation (24 images)
)

# Clean rescaling generator for validation
val_datagen = ImageDataGenerator(
    rescale=1.0 / 255.0,
    validation_split=0.2
)

train_generator = train_datagen.flow_from_directory(
    DATASET_PATH,
    target_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='training',
    seed=42
)

val_generator = val_datagen.flow_from_directory(
    DATASET_PATH,
    target_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='validation',
    seed=42
)
```

# Build & Train Models (Custom CNN vs. MobileNetV2)

### MODEL 1 : Custom CNN (Baseline) 

```python
EPOCHS = 15
custom_cnn = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(224, 224, 3)),
    MaxPooling2D(2, 2),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D(2, 2),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D(2, 2),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(3, activation='softmax')
])

custom_cnn.compile(optimizer=Adam(learning_rate=0.001), loss='categorical_crossentropy', metrics=['accuracy'])

print("--- Training Custom CNN ---")
history_cnn = custom_cnn.fit(train_generator, validation_data=val_generator, epochs=EPOCHS, verbose=1)
```

## Model 2 : MobileNetV2 Transfer Learning

```python
# 1. Enhanced Data Augmentation to help with tiny dataset scarcity
train_datagen = ImageDataGenerator(
    rescale=1.0/255.0,
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    brightness_range=[0.8, 1.2],
    horizontal_flip=True,
    vertical_flip=True,
    fill_mode='nearest',
    validation_split=0.2
)

val_datagen = ImageDataGenerator(rescale=1.0/255.0, validation_split=0.2)

train_generator = train_datagen.flow_from_directory(
    DATASET_PATH, target_size=(224, 224), batch_size=16,
    class_mode='categorical', subset='training', seed=42
)

val_generator = val_datagen.flow_from_directory(
    DATASET_PATH, target_size=(224, 224), batch_size=16,
    class_mode='categorical', subset='validation', shuffle=False, seed=42
)

# 2. Build MobileNetV2 with a frozen base
base_model = MobileNetV2(
    input_shape=(224, 224, 3), 
    include_top=False, 
    weights='imagenet'
)
base_model.trainable = False

# 3. Enhanced Classification Head for better pattern matching
model = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    Dense(256, activation='relu'),
    Dropout(0.4),
    Dense(3, activation='softmax')
])

# 4. Optimized Learning Rate (0.001 is ideal for training just the head)
model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# 5. Callbacks to capture peak performance
checkpoint = ModelCheckpoint(
    'best_mobilenetv2_rice.keras', 
    monitor='val_accuracy', 
    save_best_only=True, 
    mode='max', 
    verbose=1
)

early_stopping = EarlyStopping(
    monitor='val_accuracy', 
    patience=5, 
    mode='max', 
    restore_best_weights=True,
    verbose=1
)

# 6. Single-pass high-performance training
history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=25,
    callbacks=[checkpoint, early_stopping],
    verbose=1
)
```

# To achieve this high validation accuracy (~87%), I did the following steps:

* Faster Learning Rate: Bumped the learning rate to 0.001 so the classification head could learn effectively.

* Expanded Model Capacity: Upgraded the dense layer to 256 units with a 0.4 dropout to better map the extracted features.

* Advanced Data Augmentation: Added zoom, shear, and brightness shifts to my image generator to combat data scarcity.

* Smart Stopping: Kept the MobileNetV2 base frozen to protect its pre-trained knowledge while letting early stopping save the peak model at Epoch 7.

# MODEL COMPARISON

## 1. Custom CNN Performance: 
* Reached a peak validation accuracy of ~56.52% with higher validation loss. Because it was trained completely from scratch on a tiny dataset, it struggled to generalize and suffered from severe overfitting.

## 2. MobileNetV2 Performance:
* Reached an optimal peak validation accuracy of 86.96% with significantly lower loss (val_loss: 0.4677) by Epoch 7. By utilizing pre-trained ImageNet feature extractors, it successfully mapped the complex patterns of rice leaf diseases without breaking its foundational knowledge.

## 3. Conclusion: 
* MobileNetV2 is clearly the superior model, providing reliable, high-accuracy predictions while bypassing the erratic guessing and overfitting of the custom architecture.

# Report on Challenges Faced

## 1.  Extreme Data Scarcity: 
* Managing with only 120 total images made training a deep neural network difficult since models normally require thousands of samples.

## 2. Fine-Tuning Drops: 
* When I tried unfreezing layers to fine-tune MobileNetV2, my small dataset caused it to lose its pre-trained knowledge and drop in accuracy, which forced me to switch to a stable single-pass frozen approach.

## 3. Validation Set Volatility:
* My validation set only had 23 images, meaning a single misclassification caused my overall accuracy score to swing by over 4%.
